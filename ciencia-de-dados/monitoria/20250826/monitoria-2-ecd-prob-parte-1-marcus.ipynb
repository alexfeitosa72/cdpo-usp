#1. O que é "probabilidade"?

A teoria das probabilidades começou a ser desenvolvida formalmente no século XVI por matemáticos interessados em entender padrões em jogos de azar. Um desses matemáticos foi Girolamo Cardano (1501–1576), cuja obra não publicada "O livro dos jogos de azar" deu um pontapé inicial ao tratamento metódico da modelagem da incerteza. Acredita-se que a formalização da probabilidade como teoria matemática foi iniciada, de fato, apenas no século seguinte com os matemáticos Blaise Pascal (1623–1662) e Pierre de Fermat (1601–1665).

Considere a previsão do tempo. Suponha que estejamos interessados em saber se chove no Vaticano neste exato momento, consultamos algum serviço de previsão do tempo e vemos que há uma probabilidade de 15% de estar chovendo na cidade.

Para nós, existe uma probabilidade de 15% de que esteja chovendo no Vaticano. E quanto ao papa? Para ele ou está chovendo (probabilidade de 100%) ou não está! (probabilidade de 0%). Em toda aplicação prática, se conhecemos a realidade, então não existe o conceito de probabilidade no mundo real. Esse conceito apenas nos caracteriza uma incerteza que temos sobre o mundo. Ou seja, a probabilidade é uma forma de medir a nossa incerteza sobre os fenômenos do mundo.

## Interpretações de probabilidade

- Clássica
- Frequêntista

### Interpretação Clássica




- É baseada em eventos equiprováveis;
- Não nos diz o que fazer quando os eventos não são equiprováveis.

### Interpretação Frequêntista

$$P(A) = \lim_{n \to \infty} \frac{\text{Número de vezes que } A \text{ ocorre}}{n}$$

```python
import numpy as np
import matplotlib.pyplot as plt

n_replicas = 10000
X = []
freq = []

prob_1 = 0.9

np.random.seed(1)
for i in range(n_replicas):
  X.append(np.random.choice([0,1], 1, p = [1-prob_1,prob_1]))
  freq.append(sum(X)/len(X))

x = np.linspace(1, n_replicas, n_replicas)

plt.axhline(prob_1, linestyle='--', c='0')
plt.plot(x,freq)
plt.xlabel("Réplicas")
plt.ylabel("Frequência relativa de 1's")
plt.show()
```

Um problema com essa interpretação é justamente ela depender de repetições, por exemplo, qual a probabilidade de chover hoje? A probabilidade de chover hoje é 30% pois das vezes que o dia de hoje aconteceu, 30% das vezes choveu, isso não funciona, o dia de hoje só acontece uma vez.


### Interpretação Subjetiva

- a probabilidade representa o grau de incerteza que uma pessoa tem sobre a ocorrência de algum evento;
- depende das crenças e informações pessoais que cada pessoa possui sobre o experimento em questão;
- o exemplo da probabilidade de chover não é um problema nessa interpretação;
- porém não temos uma forma objetiva de definir a probabilidade de ocorrência de um evento de forma objetiva nessa interpretação.

# Definição Axiomática

**Axiomas de Kolmogorov** (1903–1987)

(i) A probabilidade de qualquer evento $A$ é sempre maior ou igual a zero.
$P(A) \geq 0$.
    
(ii)  A probabilidade do evento certo ($\Omega$) é 1.
$P(\Omega) = 1$.
    
(iii) Se $A$ e $B$ são eventos mutuamente exclusivos (ou seja, não podem ocorrer ao mesmo tempo), então a probabilidade de $A$ ou $B$ ocorrer é a soma das probabilidades de $A$ e $B$:
$P(A \cup B) = P(A) + P(B)$.

<!-- (iii) Se $A_1, A_2, A_3...$ é uma sequência enumerável de eventos disjuntos, então $P(\bigcup_{i=1}^\infty A_i) = \sum_{i=1}^\infty P(A_i).$-->

**Probabilidade Condicional:**
$$P(A|B) =\displaystyle\frac{P(A \cap B)}{P(B)}.$$

## Aniversariantes no mesmo dia

Dado um grupo de $N$ pessoas, qual a probabilidade de que ao menos duas delas façam aniversário em um mesmo dia?

Embora a resposta dessa pergunta pareça ser um valor baixo, dado que temos 365 dias no ano, surpreendentemente, é um valor consideravelmente alto.


Quando lidamos com probabilidades, por vezes é mais fácil encontrarmos a probabilidade do complementar do evento de interesse e esse é um dos casos em que isso ocorre. Ou seja, vamos calcular a probabilidade de que nenhuma das $N$ pessoas tenha o mesmo aniversário.

Qual seria o espaço amostral nesse exemplo?

É o espaço amostral das datas de aniversário

\begin{align*}
  \Omega = \{&\overbrace{(01/01, 01/01, 01/01, \dots, 01/01)}^{N}, \\
         & (02/01, 01/01, 01/01, \dots, 01/01), \\
         & \hspace{3.5cm}\vdots\hspace{3.5cm}, \\
         & (31/12, 31/12, 31/12, \dots, 31/12) \}
\end{align*}
Por simplicidade vamos esquecer os anos bisextos.

Supondo que todos os dias do ano são igualmente prováveis de acontecer, basta contar o número de elementos no espaço amostral e o número de elementos no evento de interesse.

O tamanho do espaço amostral $|\Omega|$, é

$$|\Omega| = \overbrace{365 \cdot 365 \cdot 365 \dots 365}^N.$$

Estamos interessados no seguinte evento

* $A^c:$ ninguém faz aniversário no mesmo dia

Vamos então contar o número de elementos nesse conjunto.

* A primeira pessoa pode fazer aniversário em qualquer dia: 365 possibilidades;
* A 2ª pessoa faz aniversário em um dia diferente da primeira: 364 possibilidades;
* A 3ª pessoa faz  aniversário em um dia diferente das duas primeiras: 363 possibilidades;
* ...
* A Nª pessoa faz aniversário em um dia diferente das N-1 primeiras: 365 - (N-1) possibilidades.

Assim, o número de elementos em $A^c$ é

\begin{equation*}
    |A^c| = 365\cdot364\cdot363\dots(365 - (N-1))
\end{equation*}

Simplificando um pouco

\begin{equation*}
    |A^c| = \frac{365!}{(365 - N)!}.
\end{equation*}

Logo, a probabilidade de não haverem empates em um grupo de $N \leq 365$ pessoas é de

$$P(A^c) = \frac{|A^c|}{|\Omega|} = \frac{365!}{365^N(365-N)!}$$

Sendo assim, a probabilidade de que exista ao menos duas pessoas com o mesmo aniversário será dada por

\begin{equation*}
    P(A) = 1 - P(A^c) = 1 - \frac{365!}{365^N(365-N)!}
\end{equation*}

Para um grupo de $N = 10$ pessoas, essa probabilidade de ao menos um empate é igual a 0.1169. Para $N=30$ essa probabilidade já é em torno de 0.7063!

```python
import numpy as np

def prob_empate_aniversario(N):
  seq = [365]*N - np.arange(N)
  return float(1 - np.exp(sum(np.log(seq))-N*np.log(365)))
```

## O problema de Monty Hall
No programa "Let's make a deal", apresentado por Monty Hall entre 1963 e 1976 e relançado diversas vezes entre 1980 e 1991, havia um jogo simples, ainda assim nada intuitivo. O jogo seguia da seguinte forma:

O participante recebe a opção de escolher uma dentre três portas. Atrás de uma delas há um carro e atrás das demais existem cabras. Uma vez feita a escolha, o apresentador do programa, que sabe o que tem atrás de cada porta, abre uma das duas portas que não foram escolhidas, revelando uma cabra. Em seguida, ele dá ao participante a oportunidade de trocar de porta.

O participante deve trocar de porta ou não?

![image.png](https://www.lavanguardia.com/files/og_thumbnail/uploads/2015/12/21/5fa297bed98fe.jpeg)

Por simplicidade, vamos considerar que o jogador sempre escolhe a porta de número 1. O resultado é o mesmo para qualquer escolha.

Para não nos perdermos, vamos começar descrevendo o espaço amostral do experimento. Perceba que temos dois experimentos acontecendo, foquemos inicialmente em qual posição o carro pode estar

$$\Omega_{carro}  = \{1,2,3\}$$

Considere o seguinte evento
* A: O carro está atrás da porta 1

Se o carro é colocado atrás de uma das portas ao acaso, é razoável assumir que
$P(A) = 1/3$, daí $P(A^c) = 1-P(A) = 2/3$.

O apresentador então abre uma das portas que não escolhemos, ou seja, a porta 2 ou 3.

$$\Omega_{apresentador} = \{2, 3\}$$

Considere o seguinte evento
* B: O apresentador abre a porta 2

Agora, se a porta 1 contém o carro, então as portas 2 e 3 contém cabras. O apresentador pode escolher qualquer uma das duas com probabilidade 1/2.
$$P(B | A) = 1/2.$$

Se o carro não está em A, existe a probabilidade de 50% que esteja na porta 2 e de 50% que esteja na porta 3. Como o apresentador nunca abriria uma porta com um carro, e existem duas opções para onde o carro poderia estar, ele ainda escolhe a porta 2 com probabilidade 1/2.

$$P(B | A^c) = 1/2$$

Qual a probabilidade de que o carro esteja em A, sabendo que o apresentador abriu a porta 2?

$$P(A | B) = \frac{P(A \cap B)}{P(B)}$$

Precisamos então encontrar $P(A\cap B)$ e $P(B)$. Pela Lei da Probabilidade Total, temos que

\begin{align*}
    P(B) &= P(B | A)P(A) + P(B | A^c) P(A^c)\\
    &= \frac{1}{2} \cdot \frac{1}{3} + \frac{1}{2} \cdot \frac{2}{3}\\
    &= \frac{1}{2}
\end{align*}

$P(A\cap B)$ pode ser encontrada de forma relativamente simples ao percerbermos que

$$P(B|A) = \frac{P(A\cap B)}{P(A)} ⇒ P(A\cap B) = P(B|A){P(A)}.$$
Então, $P(A\cap B) = P(B|A)P(A) = 1/2 \cdot 1/3 = 1/6$.

Juntando tudo vemos que
\begin{equation*}
    P(A | B) = \frac{P(A \cap B)}{P(B)} = \dfrac{1/6}{1/2} = \frac{1}{3}
\end{equation*}

Ou seja, a probabilidade de que o carro esteja atrás da porta 1 depois do apresentador revelar a porta 2 é de 1/3. Ou seja, se o participante mantiver a sua escolha, ele será vitorioso com probabilidade 1/3. Sendo assim, se o participante trocar para a terceira porta, tem uma probabilidade de 2/3 dela conter o carro.

```python

def play_monty_hall(porta_escolhida):
    print("Você escolheu a porta {}".format(porta_escolhida))
    # A porta com o prêmio é selecionada aleatóriamente
    porta_com_premio = np.random.randint(3)+1

    possibilidades = []
    for n in [1,2,3]:
        # Se o número da porta não tem o prêmio e não foi escolhida, a adiciona como uma possibilidade a ser aberta
        if(n != porta_com_premio and n != porta_escolhida):
            possibilidades.append(n)

    # Das portas disponíveis, escolhe uma para ser aberta
    porta_aberta = np.random.choice(possibilidades, size = 1)

    r = input("O apresentador abriu a porta {}. Trocar de porta? (s/n)".format(porta_aberta[0]))

    if(r == "s"):
        # Percorre as possibilidades e seleciona aquela que não foi escolhida e não foi aberta
        for n in [1,2,3]:
            if(n != porta_escolhida and n != porta_aberta):
                porta_escolhida = n
                break
        print("Porta trocada para {}".format(porta_escolhida))
    print("A porta com o prêmio era a {}.".format(porta_com_premio))
    if(porta_com_premio == porta_escolhida):
        print("Você ganhou!")
    else:
        print("Você perdeu!")
```

```python
play_monty_hall(1)
```

```python
def play_monty_hall_automatic(porta_escolhida, trocar):
    # A porta com o prêmio é selecionada aleatóriamente
    porta_com_premio = np.random.randint(3)+1

    possibilidades = []
    for n in [1,2,3]:
        # Se o número da porta não tem o prêmio e não foi escolhida, a adiciona como uma possibilidade a ser aberta
        if(n != porta_com_premio and n != porta_escolhida):
            possibilidades.append(n)

    # Das portas disponíveis, escolhe uma para ser aberta
    porta_aberta = np.random.choice(possibilidades, size = 1)

    if(trocar):
        # Percorre as possibilidades e seleciona aquela que não foi escolhida e não foi aberta
        for n in [1,2,3]:
            if(n != porta_escolhida and n != porta_aberta):
                porta_escolhida = n
    if(porta_com_premio == porta_escolhida):
        return 1
    else:
        return 0
```

```python
n_jogadas = 10000

jogos = []
for i in range(n_jogadas):
    resultado = play_monty_hall_automatic(1, True)
    jogos.append(resultado)
print("Resultados dos 20 primeiros jogos: {}".format(jogos[:20]))
print("Proporção de vitórias: {}".format(np.mean(jogos)))
```

# 3. E quando não sabemos associar a probabilidade?

Até o momento, trabalhamos com exemplos simples, em que os eventos são assumidos equiprováveis ou então nos é dita qual é a probabilidade deles. Mas e
quando isso não é possível?

No mundo real teremos uma realização de um experimento e com ele precisaremos encontrar a probabilidade. Essa é a problemática a ser explorada nos próximos módulos, a Inferência Estatística.

Mas vamos dar uma espiada no futuro.

Digamos que fizemos o lançamento de cinco moedas com faces 0 e 1, os resultados foram $1, 1, 0, 0, 1$.

A probabilidade de sair 1 nessa moeda é desconhecida, digamos que seja igual a $p$

A probabilidade de eu observar a sequência de lançamentos $1, 1, 0, 0, 1$ é de

\begin{align*}
  L(p) &= p \cdot p \cdot (1-p) \cdot (1-p) \cdot p\\
       &= p^3(1-p)^2
\end{align*}

Supomos então que a sequência que observamos é algo comum, vamos então encontrar $p$ que maximiza a chance de observarmos a sequência.


```python
import numpy as np
import matplotlib.pyplot as plt


p = np.linspace(0, 1, 1000)
L = np.pow(p,3)*np.pow(1-p, 2)

plt.plot(p,L)
plt.xlabel("Réplicas")
plt.ylabel("Frequência relativa de 1's")
plt.show()
```

```python
from scipy.optimize import minimize

def objective_function(p):
    return -p**3*(1-p)**2

initial_guess = .5
result = minimize(objective_function, initial_guess)

max_x = result.x[0]

print(f"Estimativa: {max_x:.4f}")
```
