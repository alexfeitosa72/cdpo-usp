# O Teste Qui-Quadrado


Considere a seguinte tabela de contingência (frequências absolutas observadas)

$$\begin{array}{c|cccc|c}
  \text{Variável 1 (i)} \text{ } \backslash \text{Variável 2 (j)}  & \text{Categoria } j_1 & \text{Categoria } j_2 & \cdots & \text{Categoria } j_c & \text{Total} \\
    \hline
    \text{Categoria } i_1 & O_{11} & O_{12} & \cdots & O_{1c} & O_{1\cdot} \\
    \text{Categoria } i_2 & O_{21} & O_{22} & \cdots & O_{2c} & O_{2\cdot} \\
    \vdots & \vdots & \vdots & \ddots & \vdots & \vdots \\
    \text{Categoria } i_r & O_{r1} & O_{r2} & \cdots & O_{rc} & O_{r\cdot} \\
    \hline
    \text{Total} & O_{\cdot1} & O_{\cdot2} & \cdots & O_{\cdot c} & O_{\cdot\cdot}
\end{array}$$

### Hipóteses de interesse
    
- **Hipótese Nula (H₀):** Não existe associação entre as variáveis (qualitativas).
    
- **Hipótese Alternativa (H₁):** Existe uma associação entre as variáveis (qualitativas).


Sob $H_0$ (independência das duas variáveis), temos a tabela de frequências esperadas

$$
\begin{array}{c|cccc|c}
  \text{Variável 1 (i)} \text{ } \backslash \text{Variável 2 (j)}  & \text{Categoria } j_1 & \text{Categoria } j_2 & \cdots & \text{Categoria } j_c & \text{Total} \\
    \hline
    \text{Categoria } i_1 & E_{11} & E_{12} & \cdots & E_{1c} & O_{1\cdot} \\
    \text{Categoria } i_2 & E_{21} & E_{22} & \cdots & E_{2c} & O_{2\cdot} \\
    \vdots & \vdots & \vdots & \ddots & \vdots & \vdots \\
    \text{Categoria } i_r & E_{r1} & E_{r2} & \cdots & E_{rc} & O_{r\cdot} \\
    \hline
    \text{Total} & O_{\cdot1} & O_{\cdot2} & \cdots & O_{\cdot c} & O_{\cdot\cdot}
\end{array}
$$



O valor do qui-quadrado é calculado utilizando a seguinte fórmula:

$\chi^2 = \displaystyle \sum\sum \frac{(O_{ij} - E_{ij})^2}{E_{ij}}$

onde:

- $O_{ij}$ é a frequência observada na célula correspondente à combinação das categorias $i$ e $j$ das duas variáveis.
- $E_{ij}$ é a frequência esperada na célula correspondente à combinação das categorias $i$ e $j$, calculada com base na hipótese nula.
    
    
### Passos

Para aplicar o teste qui-quadrado, seguimos os seguintes passos:

1. Formule a hipótese nula ($H_0$) e a hipótese alternativa ($H_1$). A hipótese nula geralmente afirma que não há associação entre as variáveis categóricas.
2. Calcule as frequências esperadas ($E_i$) para cada categoria, assumindo que $H_0$ é verdadeira.
3. Calcule o valor do qui-quadrado ($\chi^2$) usando a fórmula mencionada.
4. Compare o valor do qui-quadrado calculado com o valor crítico da distribuição qui-quadrado com o número apropriado de graus de liberdade para determinar se $H_0$ deve ser rejeitada.


### Exemplo: Teste Qui-Quadrado
    
#### Tabela de contingência observada: eficiência de algoritmos em dois tipos de máquinas

| Tipo de Máquina \ Eficiência | Eficiente | Ineficiente | Total |

| Computador Padrão             | 50        | 50          | 100   |
| **Total**                     | 130       | 70          | 200   |



#### Cálculo da Tabela Esperada

As frequências esperadas ($E_{ij}$) são calculadas com base na hipótese de independência:

$
E_{ij} = \displaystyle \frac{\text{Total da linha} \times \text{Total da coluna}}{\text{Total geral}}
$

#### Cálculos para cada célula:

##### Cálculo para Computador de Alto Desempenho-Eficiente

$
E_{11} =\displaystyle  \frac{(100 \times 130)}{200} = 65
$

##### Cálculo para Computador de Alto Desempenho-Ineficiente

$
E_{12} = \displaystyle \frac{(100 \times 70)}{200} = 35
$

##### Cálculo para Computador Padrão-Eficiente

$
E_{21} = \displaystyle \frac{(100 \times 130)}{200} = 65
$

##### Cálculo para Computador Padrão-Ineficiente

$
E_{22} = \displaystyle \frac{(100 \times 70)}{200} = 35
$

### Tabela Esperada

| Tipo de Máquina \ Eficiência | Eficiente | Ineficiente | Total |
|------------------------------|-----------|-------------|-------|
| Computador de Alto Desempenho | 65        | 35          | 100   |
| Computador Padrão             | 65        | 35          | 100   |
| **Total**                     | 130       | 70          | 200   |

### Cálculo da Estatística Qui-Quadrado

A estatística qui-quadrado ($\chi^2$) é dada pela fórmula:

$
\chi^2 =\displaystyle  \sum \frac{(O_{ij} - E_{ij})^2}{E_{ij}}
$

Onde:
- $O_{ij}$ são os valores observados,
- $E_{ij}$ são os valores esperados.

    
### Estatística Qui-Quadrado
$
\chi^2_{11} = \displaystyle \frac{(80 - 65)^2}{65} = \frac{(15)^2}{65} = \frac{225}{65} \approx 3.46
$

$
\chi^2_{12} = \displaystyle \frac{(20 - 35)^2}{35} = \frac{(-15)^2}{35} = \frac{225}{35} = 6.43
$

$
\chi^2_{21} = \displaystyle \frac{(50 - 65)^2}{65} = \frac{(-15)^2}{65} = \frac{225}{65} \approx 3.46
$

$
\chi^2_{22} = \displaystyle \frac{(50 - 35)^2}{35} = \frac{(15)^2}{35} = \frac{225}{35} = 6.43
$

Agora, somamos todos os valores calculados:

$
\chi^2 = 3.46 + 6.43 + 3.46 + 6.43 = 19.78
$

### Graus de Liberdade

Os graus de liberdade (gl) para o teste qui-quadrado de independência são dados por:

$
\text{gl} = (\text{número de linhas} - 1) \times (\text{número de colunas} - 1) =  (2 - 1) \times (2 - 1) = 1
$

### Conclusão do teste

Para um nível de significância $\alpha = 0.05$ e 1 grau de liberadade, o valor crítico é aproximadamente **3.84**.

Como $\chi^2 = 19.78$ é maior que o valor crítico de **3.84**, **rejeitamos a hipótese nula**. Isso significa que há evidências suficientes para concluir que **há uma associação significativa entre o tipo de máquina e a eficiência do algoritmo**.

    

### Teste Qui-quadrado: Aplicações em Python

### Aplicação 1: Eficiência de Computadores

#### Teste Qui-quadrado para independência entre variáveis usando o scipy

```python
import pandas as pd

dados = {
    'Eficiente': [80, 50],
    'Ineficiente': [20, 50],
}

index = ['Computador de Alto Desempenho', 'Computador Padrão']

df = pd.DataFrame(dados, index=index)
print(df)
```

```python
import scipy.stats as stats
import numpy as np

Q2 = stats.chi2_contingency(df)

num_linhas, num_colunas = df.shape

estatística = Q2[0]
valorp = Q2[1]

# Nível de significância do teste
alpha = 0.05
quantil = stats.chi2.ppf(1 - alpha, (num_linhas-1) * (num_colunas-1))

print('Estatística Qui-quadrado = ',np.round(estatística,3), '\nValor-p=',np.round(valorp,3), '\nQuantil Qui-quadrado = ', np.round(quantil,3))

# Conclusão do teste
if valorp < alpha:
    print("Rejeitamos a hipótese nula: Há associação entre as variáveis.")
else:
    print("Não rejeitamos a hipótese nula: Não há evidências de associação entre as variáveis.")
```

```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

# Parâmetros do teste
gl = 1  # Graus de liberdade (ajuste conforme necessário)
chi2_stat = 18.484  # Estatística do teste
chi2_critical = 3.841  # Valor crítico

# Valores da distribuição qui-quadrado
x = np.linspace(-0.1, chi2_stat + 5, 1000)
y = stats.chi2.pdf(x, gl)

# Distribuição da estatística de teste
plt.figure(figsize=(8, 3))
plt.plot(x, y, label=f'Qui-quadrado (gl={gl})', color='blue')

# Preenchendo a área crítica
x_critical = np.linspace(chi2_critical, max(x), 500)
y_critical = stats.chi2.pdf(x_critical, gl)
plt.fill_between(x_critical, y_critical, color='red', alpha=0.3, label='Região crítica')

# Linha para a estatística do teste
plt.axvline(chi2_stat, color='black', linestyle='--', label=f'Estatística do teste = {chi2_stat}')

# Linha para o valor crítico
plt.axvline(chi2_critical, color='red', linestyle='--', label=f'Valor crítico = {chi2_critical}')

# Configurações do gráfico
plt.title('Distribuição Qui-Quadrado')
plt.xlabel('Valor Qui-Quadrado')
plt.ylabel('Densidade de Probabilidade')
plt.legend()
plt.grid()
plt.show()
```

## Teste de Shapiro-Wilk para Normalidade

O teste de Shapiro-Wilk é um teste estatístico amplamente utilizado para verificar a normalidade de uma amostra. Caso o teste de Shapiro-Wilk rejeite a normalidade dos dados, testes não paramétricos são indicados para avaliar a equivalência dos grupos.
        
### Hipóteses
    
- $H_0$: A amostra foi retirada de uma população com distribuição normal.

- $H_1$: A amostra não foi retirada de uma população com distribuição normal.
  
### Estatística do Teste de Shapiro-Wilk



A estatística do teste \(W\) é definida como:

$
W =\displaystyle \frac{(\sum_{i=1}^{n} a_i x_{(i)})^2}{\sum_{i=1}^{n} (x_i - \bar{x})^2}$


onde:
- $n$ é o número de observações na amostra.
- $x_{(i)}$ é o \(i\)-ésimo valor ordenado da amostra.
- $\bar{x}$ é a média da amostra.
- $a_i$ são os coeficientes do teste, que dependem da distribuição normal.

### Interpretação da Estatística $W$

- O valor de $W$ varia entre 0 e 1.
- Valores de $W$ próximos de 1 indicam que a amostra pode ser normalmente distribuída.
- Valores de $W$ significativamente menores que 1 sugerem que a hipótese de normalidade deve ser rejeitada.
    
    
O valor-p associado à estatística $W$ é utilizado para determinar a significância do teste. Um valor-p menor que o nível de significância (tipicamente 0.05) indica que há evidências suficientes para rejeitar a hipótese nula de normalidade.


#### Formas complementares de avaliar a normalidade dos dados
    
- Histogramas;
- Gráficos de quantis.


## Teste Shapiro-Wilk - Aplicações


```python
### Implementação em Python

#Teste de Shapiro-Wilk em Python usando a biblioteca `scipy`:

import numpy as np
from scipy.stats import shapiro

# Gerar uma amostra de dados
dados = np.random.normal(loc=0, scale=1, size=100)  # 100 dados normalmente distribuídos

# Realizar o teste de Shapiro-Wilk
statistic, p_value = shapiro(dados)

# Exibir os resultados
print("Estatística do teste de Shapiro-Wilk:", statistic)
print("Valor-p:", p_value)

# Interpretar o valor-p
alpha = 0.05
if p_value < alpha:
    print("\nRejeitamos a hipótese nula: Os dados não seguem uma distribuição normal.")
else:
    print("\nNão rejeitamos a hipótese nula: Os dados seguem uma distribuição normal.")
```

## Teste de Wilcoxon
    
Suposições
    
- Os dados são pareados (emparelhados);
- Cada par é escolhido aleatoriamente e de forma independente;
- Os dados são medidos pelo menos em uma escala ordinal (i.e., eles não podem ser nominais).
    
    
Seja $n$ o tamanho da amostra, i.e., o número de pares. Assim, há um total de $2n$ observações. Para os pares $i=1, \dots, n$, $x_{1,i}$ e $x_{2,i}$ indicam as medidas.

### Hipóteses    
    
$H_{0}$: A diferença entre os pares segue uma distribuição simétrica em torno de zero
    
$H_{1}$: A diferença entre os pares não segue uma distribuição simétrica em torno de zero.

    

## Teste de Wilcoxon
    

Para $i=1,\dots, n$, calcule $|x_{2,i} - x_{1,i}|$ e $\operatorname{sgn}(x_{2,i} - x_{1,i})$, onde $\operatorname{sgn}$ é a função sinal.

Exclua os pares com $|x_{2,i} - x_{1,i}| = 0$. Seja $n_{r}$ o tamanho da amostra reduzida.

Ordene os $n_{r}$ pares remanescentes da menor diferença absoluta para a maior diferença absoluta, $|x_{2,i} - x_{1,i}|$.

Atribua postos aos pares, começando com o menor como 1. Valores repetidos recebem um posto igual à média dos postos que eles abrangem.
    
    
Seja $R_{i}$ o posto do par $i$.
    
    


### Estatística de teste: Wilcoxon

Calcule o teste estatístico $W$:
    
$$W = \sum_{i=1}^{n_{r}} [\text{sgn}(x_{2,i} - x_{1,i})R_{i}],$$

em que ${\displaystyle R_{i}}$ denota o posto do par $i$, ou seja, $W$ é a soma dos postos sinalizados.

Sob a hipótese nula, $W$ segue uma distribuição específica, sem uma expressão simples. Esta distribuição tem um valor esperado de 0 e uma variância de
    
$$\sigma_W^2 = \displaystyle \frac{n_r(n_r + 1)(2n_r + 1)}{6}.$$

$W$ pode ser comparado com um valor crítico da tabela de referência.

O teste bicaudal consiste em rejeitar $H_{0}$ se $|W| > W_{\text{crítico}, n_r}.$

Com $n_r$ grande, a distribuição amostral de $W$ se aproxima da normal.




## Teste de Mann-Whitney


- **Hipótese Nula (H₀):** As duas amostras vêm da mesma distribuição.
    
- **Hipótese Alternativa (H₁):** As duas amostras vêm de distribuições diferentes.
    


### Estatística de Teste de Mann-Whitney

A estatística de teste no Mann-Whitney $U$ é dada por $min(U_1, U_2)$, em que

$$
U_1 = n_1 n_2 + \frac{n_1(n_1 + 1)}{2} - R_1
$$

$$
U_2 = n_1 n_2 + \frac{n_2(n_2 + 1)}{2} - R_2
$$

onde:

- $n₁$ e $n₂$ são os tamanhos da primeira e da segunda amostra, respectivamente.
- $R₁$ e $R₂$ são as somas dos postos (ranks) das observações na primeira e segunda amostras, respectivamente.


A estatística de teste $U$ é comparada a uma distribuição conhecida para avaliar a significância do teste. Para amostras pequenas ($n_1, n_2 < 20$), os valores críticos são obtidos de tabelas específicas do teste de Mann-Whitney. Para amostras maiores, a estatística $U$ pode ser aproximada por uma distribuição normal:

$$Z = \frac{U - \mu_U}{\sigma_U}$$

onde:

$$\mu_U = \frac{n_1 n_2}{2}, \quad \sigma_U = \sqrt{\frac{n_1 n_2 (n_1 + n_2 + 1)}{12}}$$

Esse valor padronizado $Z$ segue aproximadamente uma distribuição normal padrão $N(0,1)$, permitindo a obtenção do p-valor para testar a hipótese nula.


## Teste de Mann-Whitney

    
    
**Exemplo**. Considere que duas amostras tenham gerado as somas de postos


$R_1$ = 1 + 3 + 4 + 6 + 8 + 12 = 34


$R_2$ = 2 + 5 + 7 + 9 + 10 + 11 = 44
     
    
$U_1 = 6 \times 6 + \displaystyle\frac{6(6 + 1)}{2} - 34 = 23$
    
$U_2 = 6 \times 6 + \displaystyle\frac{6(6 + 1)}{2} - 44 = 13$
    
$U = min(U_1,U_2) = 13$
    
Podemos aproximar a estatística $U$ por uma distribuição normal usando:

$$\mu_U = \frac{n_1 n_2}{2} = \frac{6 \times 6}{2} = 18$$

$$\sigma_U = \sqrt{\frac{n_1 n_2 (n_1 + n_2 + 1)}{12}} = \sqrt{\frac{6 \times 6 \times (6 + 6 + 1)}{12}} = \sqrt{\frac{6 \times 6 \times 13}{12}} = \sqrt{39} \approx 6.24$$

Calculamos então o valor padronizado $Z$:

$$ Z = \frac{U - \mu_U}{\sigma_U} = \frac{13 - 18}{6.24} \approx -0.80$$

##### Obtenção do p-valor

O p-valor pode ser encontrado consultando a tabela da distribuição normal padrão para $Z = -0.80$, que dá aproximadamente $P(Z \leq -0.80) \approx 0.2119$ (para um teste unilateral). Para um teste bilateral, o p-valor é o dobro desse valor, ou seja:

$$p \approx 2 \times 0.2119 = 0.4238$$

#### Conclusão

Se considerarmos um nível de significância usual $\alpha = 0.05$, o p-valor obtido ($0.4238$) é maior que $0.05$, indicando que **não há evidência suficiente para rejeitar a hipótese nula**. Assim, não podemos afirmar que as duas amostras vêm de distribuições diferentes.
  


## Teste de Kruskall-Wallis


- $H_0$: Todas as $k$ populações têm a mesma distribuição.
    
- $H_1$: Pelo menos uma das populações tendem a produzir observações maiores do que outras populações; ou, pelo menos duas das k populações diferem na localização.
    
    
Seja $R(X_{ij})$ o posto (rank) da observação de $X_{ij}$ e seja $R_i$ a soma dos postos da i-ésima amostra:

$R_i = \displaystyle \sum_{j=1}^{n_i} R(X_{ij}), i =1,\ldots,k$,

onde $n_i$ é o número de observações na i-ésima amostra e n é o total de observações.
    
    
A estatística de teste T é definida como
   
    
$T = \displaystyle\frac{1}{S^2}\left(\displaystyle \sum_{i=1}^{k} \frac{R_i^2}{n_i} - \displaystyle\frac{n(n+1)^2}{4}\right),$
em que $S^2 = \displaystyle\frac{1}{n-1}\left(\displaystyle\sum_{\forall i,j }R(X_{ij})^2 - \displaystyle\frac{n(n+1)^2}{4}\right)$.
    
Se não existe empates, a expressão de T pode ser simplificada.
    
    
A distribuição de $T$ é complexa e a conclusão do teste se baseia em valores de tabelas de teste de Kruskal-Wallis.

## Teste de Wilcoxon (duas amostras pareadas)



### Aplicação: Desempenho de máquinas antes e depois de uma atualização

```python
import numpy as np
from scipy.stats import wilcoxon
import matplotlib.pyplot as plt

# Dados de desempenho das máquinas antes e depois da atualização
desempenho_antes = np.array([85, 90, 78, 88, 92, 87, 95, 82, 80, 94, 99, 107, 130])
desempenho_depois = np.array([90, 91, 80, 89, 95, 90, 96, 85, 84, 93, 90, 100, 110])


# Criar histogramas
plt.figure(figsize=(12, 6))

# Histograma para desempenho antes
plt.subplot(1, 2, 1)
plt.hist(desempenho_antes, bins=10, color='blue', alpha=0.7)
plt.title('Desempenho Antes da Atualização')
plt.xlabel('Desempenho')
plt.ylabel('Frequência')
plt.grid(axis='y', alpha=0.75)

# Histograma para desempenho depois
plt.subplot(1, 2, 2)
plt.hist(desempenho_depois, bins=10, color='orange', alpha=0.7)
plt.title('Desempenho Depois da Atualização')
plt.xlabel('Desempenho')
plt.ylabel('Frequência')
plt.grid(axis='y', alpha=0.75)

# Exibir os histogramas
plt.tight_layout()
plt.show()
```

```python
from scipy.stats import wilcoxon, shapiro

# Teste de Shapiro-Wilk para normalidade
statistic_antes, p_value_antes = shapiro(desempenho_antes)
statistic_depois, p_value_depois = shapiro(desempenho_depois)

# Exibir os resultados do teste de Shapiro-Wilk
print("Teste de Shapiro-Wilk para desempenho antes:")
print("Estatística:", statistic_antes)
print("Valor-p:", p_value_antes)

print("\nTeste de Shapiro-Wilk para desempenho depois:")
print("Estatística:", statistic_depois)
print("Valor-p:", p_value_depois)
# Interpretar os valores-p do teste de Shapiro-Wilk
alpha = 0.05
if p_value_antes < alpha:
    print("\nRejeita a hipótese nula (desempenho antes): Os dados não são normalmente distribuídos.")
else:
    print("\nNão rejeita a hipótese nula (desempenho antes): Os dados são normalmente distribuídos.")

if p_value_depois < alpha:
    print("\nRejeito a hipótese nula (desempenho depois): Os dados não são normalmente distribuídos.")
else:
    print("\nNão rejeito a hipótese nula (desempenho depois): Os dados são normalmente distribuídos.")
```

```python
# Teste de Wilcoxon
statistic, p_value = wilcoxon(desempenho_antes, desempenho_depois)

# Exibir os resultados
print("Estatística do teste de Wilcoxon:", statistic)
print("Valor-p:", p_value)

# Interpretar o valor-p
alpha = 0.05
if p_value < alpha:
    print("\nRejeito a hipótese nula: Há uma diferença significativa no desempenho antes e depois da atualização.")
else:
    print("\nNão rejeito a hipótese nula: Não há diferença significativa no desempenho antes e depois da atualização.")
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

import warnings
warnings.filterwarnings('ignore')  # Ignora todos os warnings

%matplotlib inline

# Modifique o diretório para fazer a leitura dos dados em DadoseDecisoes.csv
url = "https://raw.githubusercontent.com/cibelerusso/Datasets/refs/heads/main/DadoseDecisoes.csv"

# Leitura dos dados
dados = pd.read_csv(url, index_col=0)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Criando o histograma de Salário por Gênero
plt.figure(figsize=(10,6))
ax = sns.histplot(data=dados, x='Salário', hue='Gênero', kde=True, bins=30, palette='Set2')

# Configurações adicionais
plt.title('Histograma de Salário por Gênero')
plt.xlabel('Salário')  # Corrigido de 'Energia' para 'Salário'
plt.ylabel('Frequência')

plt.show()
```

```python
import pandas as pd
from scipy.stats import shapiro, ttest_ind, mannwhitneyu

# Dividindo os grupos com base na variável 'Modo'
grupo1 = dados[dados['Gênero'] == 'Masculino']['Salário']
grupo2 = dados[dados['Gênero'] == 'Feminino']['Salário']

# Teste de normalidade de Shapiro-Wilk para cada grupo
shapiro_grupo1 = shapiro(grupo1)
shapiro_grupo2 = shapiro(grupo2)

print("Teste de Shapiro-Wilk para grupo 1:\n", shapiro_grupo1)
print("Teste de Shapiro-Wilk para grupo 2:\n", shapiro_grupo2)

# Avaliando se os grupos têm distribuição normal (valor p > 0.05 indica normalidade)
if shapiro_grupo1.pvalue > 0.05 and shapiro_grupo2.pvalue > 0.05:
    # Teste t de Student para igualdade das médias (para dados normais)
    t_stat, p_value = ttest_ind(grupo1, grupo2)
    print(f"Teste t de Student: t = {t_stat}, p = {p_value}")
else:
    # Teste de Mann-Whitney (para dados não normais)
    u_stat, p_value = mannwhitneyu(grupo1, grupo2)
    print(f"Teste de Mann-Whitney: U = {u_stat}, p = {p_value}");
```

## Teste de Kruskall-Wallis (mais de duas amostras não-pareadas)


```python
pd.crosstab(dados['Departamento'],columns='Total')
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Criando o histograma de Salário por Departamento
plt.figure(figsize=(10,6))
sns.histplot(data=dados, x='Salário', hue='Departamento', kde=True, bins=30, palette='Set2')

# Configurações adicionais
plt.title('Histograma de Salário por Departamento')
plt.xlabel('Salário')
plt.ylabel('Frequência')


# Exibindo o gráfico
plt.show()
```

```python
import pandas as pd
from scipy.stats import shapiro, kruskal

# Lista para armazenar os resultados do teste de Shapiro-Wilk por Compasso
shapiro_results = {}

# Teste de Shapiro-Wilk por grupos de Compasso
for departamento in dados['Departamento'].unique():
    grupo = dados[dados['Departamento'] == departamento]['Salário']
    stat, p_value = shapiro(grupo)
    shapiro_results[departamento] = p_value
    print(f"Teste de Shapiro-Wilk para Departamento {departamento}: p = {p_value}");
```

```python
# Se algum grupo não seguir uma distribuição normal, usaremos o teste de Kruskal-Wallis

if all(p > 0.05 for p in shapiro_results.values()):
    print("Todos os grupos seguem uma distribuição normal. Um teste paramétrico poderia ser usado.")
else:
    # Realizando o teste de Kruskal-Wallis para comparar os grupos
    grupos = [dados[dados['Departamento'] == departamento]['Salário'] for departamento in dados['Departamento'].unique()]
    stat, p_value = kruskal(*grupos)
    print(f"Teste de Kruskal-Wallis: H = {stat}, p = {p_value}")
```

```python
# Conclusão
alpha = 0.05
if p_value < alpha:
    print("Conclusão: Há evidências suficientes para rejeitar a hipótese nula.\n As distribuições de 'Salário' são significativamente diferentes entre os grupos de 'Departamento'.")
else:
        print("Conclusão: Não há evidências suficientes para rejeitar a hipótese nula.\n As distribuições de 'Salário' não são significativamente diferentes entre os grupos de 'Departamento'.")
```
