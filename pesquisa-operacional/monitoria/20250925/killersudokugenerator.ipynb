```python
# Use origem O = (i,j) + WASD para criar a área relativa a O
raw_cages = [
    ((0, 0), "D", 3),  
    ((0, 2), "DD", 15),  
    ((0, 5), "SAS", 22),  
    ((0, 6), "S", 4),  
    ((0, 7), "S", 16),  
    ((0, 8), "SSS", 15),  
    ((1, 0), "DSA", 25),  
    ((1, 2), "D", 17),  
    ((2, 2), "DS", 9),  
    ((2, 5), "SS", 8),  
    ((2, 6), "DAS", 20),  
    ((3, 0), "S", 6),  
    ((3, 1), "D", 14),  
    ((3, 4), "SS", 17),  
    ((3, 7), "SA", 17),  
    ((4, 1), "DAS", 13),  
    ((4, 3), "SS", 20),  
    ((4, 8), "S", 12),  
    ((5, 0), "SSS", 27),  
    ((5, 2), "SA", 6),  
    ((5, 5), "SD", 20),  
    ((5, 6), "D", 6),  
    ((6, 4), "SAS", 10),  
    ((6, 7), "SDW", 14),  
    ((7, 1), "S", 8),  
    ((7, 2), "S", 16),  
    ((7, 5), "D", 15),  
    ((8, 4), "DD", 13),  
    ((8, 7), "D", 17)
]

def cage_processing(cage_descriptor):
    """
    Converte um descritor ((i, j), "SADW...", soma) em {'cells': [...], 'sum': soma},
    seguindo as direções relativas à última célula adicionada (tipo snake movement),
    e eliminando células repetidas.
    """
    (i, j), directions, soma = cage_descriptor
    moves = {'S': (1, 0), 'A': (0, -1), 'W': (-1, 0), 'D': (0, 1)}
    
    visited = set()
    cells = []

    def add_cell(ci, cj):
        if (ci, cj) not in visited:
            visited.add((ci, cj))
            cells.append((ci, cj))
    
    # Adiciona a célula de origem
    add_cell(i, j)
    
    # Caminha célula por célula, atualizando a referência a cada passo
    ci, cj = i, j  # última célula adicionada
    for d in directions:
        if d not in moves:
            raise ValueError(f"Direção inválida: '{d}'")
        di, dj = moves[d]
        ci, cj = ci + di, cj + dj
        add_cell(ci, cj)
    
    return {'cells': cells, 'sum': soma}

cages = [cage_processing(desc) for desc in raw_cages]

cages
```
