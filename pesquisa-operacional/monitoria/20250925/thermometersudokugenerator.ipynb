```python
# Use Origin + QWEASDZXC notation for diagonals + cardinal directions to build instance
raw_thermometers = [
    ((0, 1), "DDD", ),  
    ((0, 5), "DDXXX", ),  
    ((1, 2), "XX", ),  
    ((3, 6), "AWW", ),  
    ((4, 0), "XXX", ),  
    ((4, 8), "ZQ", ),  
    ((5, 5), "QZXX", ),  
    ((7, 1), "EWQ", ),  
    ((7, 5), "WA", ),  
    ((7, 7), "W", )
]

def thermometer_processing(thermometer_descriptor):
    """
    Converte um descritor ((i, j), "SADW...", soma) em {'cells': [...], 'sum': soma},
    seguindo as direções relativas à última célula adicionada (tipo snake movement),
    e eliminando células repetidas.
    """
    (i, j), directions = thermometer_descriptor
    moves = {
        'Q': (-1, -1), 
        'W': (-1, 0), 
        'E': (-1, 1), 
        'A': (0, -1),
        'S': (0, 0),
        'D': (0, 1),
        'Z': (1, -1),
        'X': (1, 0),
        'C': (1, 1)
    }

    # Adiciona a célula de origem
    cells = [(i, j)]
    
    # Caminha célula por célula, atualizando a referência a cada passo
    ci, cj = i, j  # última célula adicionada
    for d in directions:
        if d not in moves:
            raise ValueError(f"Direção inválida: '{d}'")
        di, dj = moves[d]
        ci, cj = ci + di, cj + dj
        cells.append((ci, cj))
    
    return cells

thermometers = [thermometer_processing(desc) for desc in raw_thermometers]

thermometers
```
