```python
gradf_numerical = sp.lambdify( ( x, y ), gradf, "numpy" )
hessf_numerical = sp.lambdify( ( x, y ), hessf, "numpy" )

maxiter = 10000
p = np.array([ 1, 0 ])
alpha = 1e-3

for i in range( maxiter ):
    p = p - alpha * np.array( gradf_numerical( *p ), dtype=float ).flatten()
    
print( p )
print( gradf_numerical( *p ).flatten() )
print( np.format_float_scientific( np.linalg.norm( gradf_numerical( *p ).flatten() ) ))
```

```python
maxiter = 100000
p = np.array([ 1, 0 ])
alpha = 1e-3

for i in range( maxiter ):
    H = np.array( hessf_numerical( *p ), dtype=float )
    G = np.array( gradf_numerical( *p ), dtype=float ).flatten()
    d = np.linalg.solve( H, G )
    p = p - alpha * d

print( p )
print( gradf_numerical( *p ).flatten() )
print( np.linalg.norm( gradf_numerical( *p ).flatten() ) )
```

```python
p = np.array([1.0, 0.0])  
v = np.zeros_like(p)      
alpha = 1e-3              
beta = 0.9                
maxiter = 100000

for i in range( maxiter ):
    lookahead = p - beta * v
    grad = np.array( gradf_numerical( *lookahead ), dtype=float ).flatten()
    v = beta * v + alpha * grad
    p = p - v

print( p )
print( gradf_numerical( *p ).flatten() )
print( np.format_float_scientific(np.linalg.norm( gradf_numerical( *p ).flatten() ) ))
```
