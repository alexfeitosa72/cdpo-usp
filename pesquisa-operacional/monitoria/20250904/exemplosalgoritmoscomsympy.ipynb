```python
import sympy as sp
from IPython.display import display, Math  
x, y  = sp.symbols('x y')
f = ( 1 - x ) ** 2 + ( y - x ** 2 ) ** 2
variables = sp.Matrix([x, y])
gf = f.diff(variables)
gf_ = sp.lambdify( variables, gf )
```

```python
hf = f.diff( variables ).jacobian( variables )
hf_ = sp.lambdify( variables, hf )

p = np.array([0, 0])
alpha = 0.1
for i in range(1000):
    # H * d = G
    d = np.linalg.solve( hf_( *p ), gf_( *p ).flatten())
    p = p - alpha * d 

print( p )
```

```python
f_ = sp.lambdify( variables, f )
gf_ = sp.lambdify( variables, gf )

p = np.array([0, 0])
q_prev = np.array([0, 0])

t_prev = 1
step = 1e-1
for i in range( 100 ):
    q = p - step * gf_( *p ).flatten()
    t = 0.5 * ( 1 + np.sqrt( 1 + 4 * t_prev ** 2 ) )
    p = q + ( t_prev - 1 ) / t * ( q - q_prev )
    t_prev = t
    q_prev = q

print( p )
```
