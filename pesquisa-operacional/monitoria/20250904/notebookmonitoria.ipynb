```python
from sympy import symbols
from IPython.display import display, Math
import sympy as sp

x, y = symbols('x y') 

f = ( 1 - x ) ** 2 + ( y - x**2 ) ** 2
```

```python
display(Math(rf"""
    \begin{{array}}{{lc}}
        f(x,y) & = {sp.latex(f)}\\
        f(2,y) & = {sp.latex(f.subs(x, 2))}\\
        f(2,3) & = {sp.latex(f.subs(x, 2).subs(y, 3))}\\
    \end{{array}}
"""))
```

```python
f.diff( x )
```

```python
f.diff( y ).integrate( x )
```

```python
variables = sp.Matrix([x, y])
f.diff( variables )


```
