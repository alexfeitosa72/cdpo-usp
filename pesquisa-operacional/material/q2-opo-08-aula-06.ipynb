## 1. Conjuntos convexos

Dizemos que um conjunto $C \subset \mathbb R^n$ é convexo se quando dados dois pontos quaisquer
$$\def\vect{\boldsymbol}
    \vect x, \vect y \in C,
$$
todo o segmento de reta ligando $\vect x$ a $\vect y$ está contido em $C$.

Ou seja,
$$
    \vect x, \vect y \in C \Rightarrow \alpha \vect x + ( 1 - \alpha ) \vect y \in C\quad \forall \alpha \in [ 0, 1 ].
$$

## 2. Funções Convexas

Dizemos que uma função $f : \mathbb R^n \to \mathbb R$ é convexa quando, para quaisquer $\vect x, \vect y \in \mathbb R^n$ e $\alpha \in [ 0 ,1 ]$, temos
$$
    f\bigl( \alpha \vect x + ( 1 - \alpha ) \vect y \bigr) \le \alpha f( \vect x ) + ( 1 - \alpha ) f( \vect y ).
$$

Isto significa que o segmento de reta unindo dois pontos no gráfico da função nunca está abaixo do gráfico.

```python
x = np.linspace( -1.5, 1.5, 100 )
y = x ** 2

np.random.seed( 0 )

pp.figure( figsize = ( 20, 20 ) )

for i in range( 1, 5 ):
    pp.subplot( 2, 2, i )
    pp.plot( x, y )
    pts = np.random.uniform( low = -1.5, high = 1.5, size = 2 )
    while abs( pts[ 0 ] - pts[ 1 ] ) < 1.5:
        pts = np.random.uniform( low = -1.5, high = 1.5, size = 2 )
    pp.scatter( pts[ 0 ], pts[ 0 ] ** 2, c = 'k', zorder = 10 )
    pp.scatter( pts[ 1 ], pts[ 1 ] ** 2, c = 'k', zorder = 10 )
    pp.plot( pts, pts ** 2, c = 'k' )

pp.show()
```

```python
x = np.linspace( -1.5, 1.5, 100 )
y = np.exp( x )

np.random.seed( 0 )

pp.figure( figsize = ( 20, 20 ) )

for i in range( 1, 5 ):
    pp.subplot( 2, 2, i )
    pp.plot( x, y )
    pts = np.random.uniform( low = -1.5, high = 1.5, size = 2 )
    while abs( pts[ 0 ] - pts[ 1 ] ) < 1.5:
        pts = np.random.uniform( low = -1.5, high = 1.5, size = 2 )
    pp.scatter( pts[ 0 ], np.exp( pts[ 0 ] ), c = 'k', zorder = 10 )
    pp.scatter( pts[ 1 ], np.exp( pts[ 1 ] ), c = 'k', zorder = 10 )
    pp.plot( pts, np.exp( pts ), c = 'k' )

pp.show()
```

Existem outras formas de reconhecer uma função convexa:

- Uma função diferenciável é convexa se e somente se, para todos vetores $\vect x$ e $\vect y$, temos
$$
    f( \vect y ) \ge f( \vect x ) + \nabla f( \vect x )^T( \vect y - \vect x )
$$

- Uma função duas vezes diferenciável é convexa se e somente se, para todos vetores $\vect x$ e $\vect y$, temos
$$
    \vect y^T\nabla^2 f( \vect x )\vect y \ge 0
$$

- O máximo entre funções convexas é uma função convexa
$$
    g( \vect x ) = \max \{ f_1( \vect x ), f_2( \vect x ) \}
$$

- Funções afim (lineares mais uma constante) são convexas

- O produto de uma função convexa por um escalar não-negativo é uma função convexa

- A soma de funções convexas é uma função convexa

## 3. Otimização Convexa

A forma geral de um problema de otimização não-linear é
$$\def\vect{\boldsymbol}
    \begin{align*}
        \min       & \quad f( \vect x )\\
        \text{s.a:}& \quad g_i( \vect x ) \le 0 & i \in \{ 1, 2, \dots, p \}\phantom.\\
                   & \quad h_j( \vect x ) = 0   & j\in \{ 1, 2, \dots, q \}.
    \end{align*}
$$

Com
$$\def\vect{\boldsymbol}
    \begin{align*}
        &f( \vect x ) : \mathbb R^n \to \mathbb R &\\
        &g_i( \vect x ) : \mathbb R^n \to \mathbb R &\quad i \in \{ 1, 2, \dots, p \}\phantom.\\
        &h_j( \vect x ) : \mathbb R^n \to \mathbb R &\quad j \in \{ 1, 2, \dots, q \}.        
    \end{align*}
$$

Dizemos que um problema de otimização não-linear é convexo quando as funções $f$ e $g_i$ são convexas e as funções $h_j$ são funções afins (lineares mais uma constante).

Ou seja, um problema de otimização convexa tem a forma
$$\def\vect{\boldsymbol}
    \begin{align*}
        \min       & \quad f( \vect x )\\
        \text{s.a:}& \quad g_i( \vect x ) \le 0 & i \in \{ 1, 2, \dots, p \}\phantom.\\
                   & \quad A\vect x = \vect b.
    \end{align*}
$$
com as funções envolvidas convexas.

- Se $g$ é uma função convexa, então o conjunto da forma
$$
    \{ \vect x : g( \vect x ) \le 0 \}
$$
é convexo

- Conjuntos da forma
$$
    \{ \vect x : A\vect x = \vect b \}
$$
são convexos

- Em otimização convexa, todo minimizador é global

- Existem algoritmos mais eficientes para otimização convexa

<h2>1. O problema</h2>

Vamos considerar um problema da forma
$$\def\vect{\boldsymbol}
    \min_{\vect x \in \mathbb R^n} \quad f( \vect x ),
$$
onde $f : \mathbb R^n \to \mathbb R$ é convexa e diferenciável.

Ademais, supomos que o gradiente de $f$ é Lipschitz. Isto é, existe $L_f \ge 0$ tal que
$$
    \| \nabla f( \vect x ) - \nabla f( \vect y ) \|_2 \le L_f\| \vect x - \vect y \|_2.
$$

<h3>1.1. Quadrados Mínimos</h3>

Como exemplo, resolveremos um problema de quadrados mínimos da forma
$$
    \min_{\vect x \in \mathbb R^n} \quad \frac12 \| A\vect x - \vect b \|_2^2.
$$

Neste caso, temos
$$
    f( \vect x ) = \frac12 \| A\vect x - \vect b \|_2^2,
$$
$$
    \nabla f( \vect x ) = A^T( A\vect x - \vect b )
$$
e
$$
    \nabla^2 f( \vect x ) = A^TA.
$$

Como a hessiana $\nabla^2 f( \vect x )$ é semi-definida positiva, $f$ é convexa. Além disso, observe que
\begin{align*}
    \| \nabla f( \vect x ) - \nabla f( \vect y ) \|_2 &{}= \| A^TA( \vect x - \vect y ) \|_2\\
        &{}\le \|A^TA\|_2 \| \vect x - \vect y \|_2.
\end{align*}

<h2>2. Máxima Descida</h2>

Métodos de primeira ordem utilizam apenas informações do gradiente e do valor da função objetivo. Muitas vezes, para problemas com um grande número $n$ de variáveis, esta informação é tudo que podemos calcular em um tempo razoável.

Por exemplo, para este tipo de problema, o algoritmo de máxima descida com passo fixo
$$
    \vect x^{( k + 1 )} = \vect x^{ ( k ) } - \lambda \nabla f( \vect x^{ ( k ) } )
$$
é convergente se $\lambda > 0$ é suficientemente pequeno (mais precisamente, para $\lambda < 2 / L_f$).

De fato, a convexidade não é necessária para este método funcionar.

<h3>2.1. Quadrados Mínimos</h3>

Para o problema de quadrados mínimos
$$
    \min_{\vect x \in \mathbb R^n} \quad \frac12 \| A\vect x - \vect b \|_2^2.
$$

O método
$$
    \vect x^{( k + 1 )} = \vect x^{ ( k ) } - \lambda \nabla f( \vect x^{ ( k ) } )
$$
assume, portanto, a forma
$$
    \vect x^{( k + 1 )} = \vect x^{ ( k ) } - \lambda A^T( A\vect x^{ ( k ) } - \vect b ).
$$

<h3>2.2. Aplicação</h3>

Neste exemplo,  os dados $\vect b$ serão os pixels de uma versão borrada de uma imagem desconhecida e desejamos obter a imagem original a partir da sua versão borrada.

O operador $A$ é tal que se $\vect x$ é a imagem original, então $A\vect x$ é a versão borrada. Esta operação $\vect x \mapsto A\vect x$ é obtida através da _convolução_ da imagem com uma imagem constante, chamada de _kernel_ da convolução.

A adjunta $A^T$ da convolução é a _correlação_. Ambas operações estão implementadas no módulo `scipy.signal` através das funções `convolve` e `correlate`.

Principiamos por nossa imagem inicial, que é a foto de uma astronauta. Neste exemplo, usaremos uma versão monocromática da imagem original colorida. Esta imagem está disponível no módulo `skimage`:

```python
import skimage as si              # Para os dados da imagem
import skimage.color as sc        # Para converter para monocromática
import matplotlib.pyplot as pp    # Para mostrar imagem e gráficos
import numpy as np                # Para rotinas numéricas
import scipy.signal as ss         # Para convolução e correlação

# Astronauta monocromática
img = sc.rgb2gray( si.data.astronaut() )

# O número de variáveis do problema
# é o número de pixels da imagem
print( np.prod( img.shape ) )

pp.figure( figsize = ( 6, 6 ) )
pp.imshow( img, cmap = pp.cm.gray )
pp.axis( 'off' )
pp.show()
```

Agora iremos implementar as operações $\vect x \mapsto A\vect x$ e $\vect x \mapsto A^T\vect x$ em termos das funções `convolve` e `correlate`:

```python
# Atenção, são adjuntas somente se
# o tamanho do kernel é ímpar.
N_kernel = 7
kernel = np.full( ( N_kernel, N_kernel ), 1 / ( N_kernel ** 2 ) )

def A( x ):

    return ss.convolve( x, kernel, mode = 'same' )

def AT( x ):

    return ss.correlate( x, kernel, mode = 'same' )
```

Testa se $\vect y^T A \vect x = \vect x^T A^T\vect y:$

```python
x = np.random.standard_normal( size = img.shape )
y = np.random.standard_normal( size = img.shape )

print( ( y * A( x ) ).sum() )
print( ( x * AT( y ) ).sum() )
```

Aqui criamos os dados, que são uma versão borrada da imagem original:

```python
# Cria imagem borrada:
b = A( img )

pp.figure( figsize = ( 12, 6 ) )

pp.subplot( 1, 2, 1 )
pp.imshow( img, cmap = pp.cm.gray )
pp.title( 'Original' )
pp.axis( 'off' )

pp.subplot( 1, 2, 2 )
pp.imshow( b, cmap = pp.cm.gray )
pp.title( 'Borrada' )
pp.axis( 'off' )

pp.show()
```

```python
def f( x ):

    return 0.5 * ( ( A( x ) - b ) ** 2 ).sum()

def grad( x ):

    return AT( A( x ) - b )
```

```python
def md( x, grad, step, niter ):

    iters = [ x ]

    for _ in range( niter ):

        x = x - step * grad( x )
        iters.append( x )

    return iters
```

```python
niter = 100

iters = md( b, grad, 1.0, niter )
```

```python
f_vals = [ f( x ) for x in iters ]
```

```python
pp.figure( figsize = ( 12, 12 ) )

pp.subplot( 2, 2, 1 )
pp.imshow( img, cmap = pp.cm.gray )
pp.title( 'Original' )
pp.axis( 'off' )

pp.subplot( 2, 2, 2 )
pp.imshow( iters[ -1 ], cmap = pp.cm.gray )
pp.title( 'Reconstrução' )
pp.axis( 'off' )

pp.subplot( 2, 1, 2 )
pp.semilogy( f_vals )
pp.xlabel( 'Iteração' )
pp.ylabel( 'Função objetivo' )
pp.axis( 'tight' )

pp.show()
```

<h2>3. O Método de Nesterov</h2>

Para funções convexas com gradiente Lipschitz, o método de Nesterov é mais apropriado.

nesterov_1983.png

Com uma notação mais próxima da atualmente utilizada, o método é dado por
$$
\begin{align*}
    \vect y^{( k )} &{}= \vect x^{( k )} - \lambda \nabla f( \vect x^{( k )} )\\
    t_{k + 1} &{}= \frac{1 + \sqrt{1 + 4t_k^2}}2\\
    \vect x^{( k + 1 )} &{}= \vect y^{( k )} + \frac{t_k - 1}{t_{k + 1}}( \vect y^{( k )} - \vect y^{( k - 1 )} )
\end{align*}
$$
a inicialização é feita com algum $\vect x^{( 0 )}$ escolhido pelo usuário, $k = 0$, $t_0 = 1$ e $\vect y^{( -1 )} = \vect x^{( 0 )}$.

Abaixo temos um exemplo de implementação deste algoritmo:

```python
def nesterov( x, grad, step, niter ):

    iters = [ x ]

    y_prev = x
    t = 1

    for _ in range( niter ):

        y = x - step * grad( x )

        t_prev = t
        t = 0.5 * ( 1 + ( 1 + 4 * t ** 2 ) ** 0.5 )

        x = y + ( t_prev - 1 ) / t * ( y - y_prev )

        y_prev = y

        iters.append( x )

    return iters
```

```python
iters_n = nesterov( b, grad, 1.0, niter )
```

```python
f_vals_n = [ f( x ) for x in iters_n ]
```

```python
pp.figure( figsize = ( 12, 18 ) )

pp.subplot( 3, 2, 1 )
pp.imshow( img, cmap = pp.cm.gray )
pp.title( 'Original' )
pp.axis( 'off' )

pp.subplot( 3, 2, 2 )
pp.imshow( iters_n[ -1 ], cmap = pp.cm.gray )
pp.title( 'Reconstrução' )
pp.axis( 'off' )

pp.subplot( 3, 1, 2 )
pp.semilogy( f_vals )
pp.semilogy( f_vals_n )
pp.legend( [ 'Gradiente', 'Nesterov' ] )
pp.xlabel( 'Iteração' )
pp.ylabel( 'Função objetivo' )
pp.axis( 'tight' )

pp.subplot( 3, 2, 5 )
pp.imshow( iters[ -1 ], cmap = pp.cm.gray )
pp.title( 'Gradiente' )
pp.axis( 'off' )

pp.subplot( 3, 2, 6 )
pp.imshow( iters_n[ -1 ], cmap = pp.cm.gray )
pp.title( 'Nesterov' )
pp.axis( 'off' )

pp.show()
```

## 1. Otimização Quadrática Convexa

### 1.1. Funções Quadráticas Convexas

Dizemos que uma função $f : \mathbb R^n \to \mathbb R$ é quadrática quando tem a forma
$$\def\vect{\boldsymbol}
    f( \vect x ) = \frac12\vect x^TQ\vect x + \vect c^T\vect x,
$$
onde $\vect c\in\mathbb R^n$ e $Q \in \mathbb R^{n \times n}$ simétrica.

O gradiente de uma quadrática é dado por
$$
    \nabla f( \vect x ) = Q\vect x + \vect c
$$
e a hessiana é simplesmente
$$
    \nabla^2 f( \vect x ) = Q.
$$

Quando $Q$ for semi-definida positiva, além de simétrica, $f$ é convexa.

### 1.2. Otimização Quadrática Convexa

Dizemos que um problema de otimização não-linear da forma
$$\def\vect{\boldsymbol}
    \begin{align*}
        \min       & \quad f( \vect x )\\
        \text{s.a:}& \quad g_i( \vect x ) \le 0 & i \in \{ 1, 2, \dots, p \}\\
    \end{align*}
$$
é um problema de otimização quadrática quando todas as funções envolvidas forem quadráticas.

- Se, além de quadráticas, todas as funções forem convexas, é um problema de otimização quadrática convexa.

- Observe que otimização linear é um caso especial de otimização quadrática.

### 1.3. Otimização Quadrática Convexa sem Restrições

Se nosso problema tiver a forma
$$
    \min\quad f( \vect x ) = \frac12\vect x^TQ\vect x + \vect c^T\vect x,
$$
e $Q$ for semi-definida positiva, então sua solução pode ser obtida através da condição
$$
    \nabla f( \vect x ) = \vect 0,
$$
que equivale a
$$
    Q\vect x = -\vect c.
$$

- Se $Q$ não for semi-definida positiva, o problema de minimizar $f$ sem restrições é ilimitado

- Se o problema possuir muitas variáveis, não é possível resolver o sistema linear acima diretamente

## 2. Quadrados Mínimos Como uma Função Quadrática

A função de quadrados mínimos é quadrática e convexa, como vemos a seguir
$$\def\vect{\boldsymbol}
    \begin{align*}
        \frac12\| A\vect x - \vect b \|_2^2 &{}= \frac12 ( A\vect x - \vect b )^T( A\vect x - \vect b )\\
            &{}=\frac12\vect x^TA^TA\vect x - \vect b^TA\vect x + \frac12\vect b^T\vect b.
    \end{align*}
$$

- na otimização, podemos ignorar o termo constante $\frac12\vect b^T\vect b$

- fazemos $Q = A^TA$ (veja que esta matriz é simétrica e semi-definida positiva)

- fazemos $\vect c = -A^T\vect b$

- a solução satisfaz, portanto,
$$
    A^TA\vect x = A^T\vect b.
$$

## 3. Método dos gradientes conjugados

Aqui, iremos supor que $Q$ é simétrica definida positiva (SPD). Uma matriz $Q \in \mathbb R^{n \times n}$ é dita definida positiva quando
$$
    \vect x^T Q \vect x >0 \quad \forall \vect x \neq \vect 0.
$$

No caso de uma quadrática com matriz SPD, podemos usar o método dos gradientes conjugados, dado por
$$
    \begin{align*}
        \lambda_k &{}= -\frac{\nabla f( \vect x^{( k )} )^T\vect d^{( k )}}{( \vect d^{( k )} )^TQ\vect d^{( k )}}\\[1.5em]
        \vect x^{( k + 1 )} &{}= \vect x^{( k )} + \lambda_k \vect d^{( k )}\\[1em]
        \beta_k &{}= \frac{\nabla f( \vect x^{( k + 1 )} )^T\nabla f( \vect x^{( k + 1 )} )}{\nabla f( \vect x^{( k )} )^T\nabla f( \vect x^{( k )} )}\\[1.5em]
        \vect d^{( k + 1 )} &{} = -\nabla f( \vect x^{( k + 1 )} ) + \beta \vect d^{( k )}
    \end{align*}
$$

- incializamos $\vect d^{( 0 )} = -\nabla f( \vect x^{( 0 )} )$

- o método dos gradientes conjugados converge em $n$ iterações
    - não usamos as $n$ iterações na prática (é um método iterativo)

    - erros de arredondamento causam dificuldades ao algoritmo

- o cálculo do gradiente da próxima iteração pode ser feito através de
$$
    \nabla f( \vect x^{( k + 1 )} ) = \nabla f( \vect x^{( k )} ) + \lambda_kQ\vect d^{( k )}
$$

- com esta forma de calcular, o método usa apenas um produto matriz-vetor por iteração

- o tamanho de passo é o que minimiza $f$ na reta $\vect x^{( k )} + \lambda \vect d^{( k )}$

- o método dos gradientes conjugados está implementado em `scipy.sparse.linalg.cg`

## 4. Exemplo

```python
import skimage as si              # Para os dados da imagem
import skimage.color as sc        # Para converter para monocromática
import matplotlib.pyplot as pp    # Para mostrar imagem e gráficos
import numpy as np                # Para rotinas numéricas
import scipy.signal as ss         # Para convolução e correlação

# Astronauta monocromática
img = sc.rgb2gray( si.data.astronaut() )
```

```python
# Atenção, são adjuntas somente se
# o tamanho do kernel é ímpar.
N_kernel = 7
kernel = np.full( ( N_kernel, N_kernel ), 1 / ( N_kernel ** 2 ) )

def A( x ):

    return ss.convolve( x, kernel, mode = 'same' )

def AT( x ):

    return ss.correlate( x, kernel, mode = 'same' )
```

```python
# Cria imagem borrada:
b = A( img )
```

```python
def f( x ):

    return 0.5 * ( ( A( x ) - b ) ** 2 ).sum()

def grad( x ):

    return AT( A( x ) - b )
```

```python
def md( grad, x, l, n ):

    iters = [ x ]

    for i in range( n ):

        x = x - l * grad( x )
        iters.append( x )

    return iters
```

```python
niter = 100

iters = md( grad, b, 1.0, niter )
```

```python
f_vals = [ f( x ) for x in iters ]
```

```python
def nesterov( x, grad, step, niter ):

    iters = [ x ]

    y_prev = x
    t = 1

    for _ in range( niter ):

        y = x - step * grad( x )

        t_prev = t
        t = 0.5 * ( 1 + ( 1 + 4 * t ** 2 ) ** 0.5 )

        x = y + ( t_prev - 1 ) / t * ( y - y_prev )

        y_prev = y

        iters.append( x )

    return iters
```

```python
iters_n = nesterov( b, grad, 1.0, niter )
```

```python
f_vals_n = [ f( x ) for x in iters_n ]
```

### 4.1. Resolvendo com o Método dos Gradientes Conjugados

```python
def cg( Q, c, x, N ):

    iters = [ x ]

    g = Q( x ) + c
    d = -g

    for k in range( N ):

        Qd = Q( d )
        l = -( g * d ).sum() / ( d * Qd ).sum()

        x = x + l * d

        iters.append( x )

        g_next = g + l * Qd
        beta = ( g_next * g_next ).sum() / ( g * g ).sum()

        d = -g_next + beta * d
        g = g_next

    return iters
```

```python
iters_cg = cg( lambda x : AT( A( x ) ), -AT( b ), b, niter )
```

```python
f_vals_cg = [ f( x ) for x in iters_cg ]
```

```python
pp.figure( figsize = ( 18, 27 ) )

pp.subplot( 3, 2, 1 )
pp.imshow( img, cmap = pp.cm.gray )
pp.title( 'Original' )
pp.axis( 'off' )

pp.subplot( 3, 2, 2 )
pp.imshow( iters_cg[ -1 ], cmap = pp.cm.gray )
pp.title( 'Reconstrução' )
pp.axis( 'off' )

pp.subplot( 3, 1, 2 )
pp.semilogy( f_vals )
pp.semilogy( f_vals_n )
pp.semilogy( f_vals_cg )
pp.legend( [ 'Gradiente', 'Nesterov', 'Gradientes Conjugados' ] )
pp.xlabel( 'Iteração' )
pp.ylabel( 'Função objetivo' )
pp.axis( 'tight' )

pp.subplot( 3, 2, 5 )
pp.imshow( iters_n[ -1 ], cmap = pp.cm.gray )
pp.title( 'Nesterov' )
pp.axis( 'off' )

pp.subplot( 3, 2, 6 )
pp.imshow( iters_cg[ -1 ], cmap = pp.cm.gray )
pp.title( 'Gradientes Conjugados' )
pp.axis( 'off' )

pp.show()
```

### 4.2. Resolvendo com o Método de Máxima Descida com Passo Exato

```python
def mdpe( Q, c, x, n ):

    iters = [ x ]

    g = Q( x ) + c

    for i in range( n ):

        Qd = Q( g )
        l = -( g * g ).sum() / ( g * Qd ).sum()

        x = x + l * g

        g = g + l * Qd

        iters.append( x )

    return iters
```

```python
iters_mdpe = mdpe( lambda x : AT( A( x ) ), -AT( b ), b, niter )
```

```python
f_vals_mdpe = [ f( x ) for x in iters_mdpe ]
```

```python
pp.figure( figsize = ( 18, 9 ) )

pp.semilogy( f_vals )
pp.semilogy( f_vals_mdpe )
pp.semilogy( f_vals_n )
pp.semilogy( f_vals_cg )
pp.legend( [ 'Gradiente', 'Gradiente Passo Exato', 'Nesterov', 'Gradientes Conjugados' ] )
pp.xlabel( 'Iteração' )
pp.ylabel( 'Função objetivo' )
pp.axis( 'tight' )

pp.show()
```

<h2>1. O Problema</h2>

Vamos considerar um problema da forma
$$
    \def\vect{\boldsymbol}
    \def\prox{\mathrm{prox}}
    \def\argmin{\mathop{\mathrm{arg\,min}}}
    \min_{\vect x \in \mathbb R^n} \quad F( \vect x ) = f( \vect x ) + g( \vect x ),
$$
onde $f : \mathbb R^n \to \mathbb R$ é convexa e diferenciável e $g : \mathbb R^n \to \mathbb R$ é convexa.

Ademais, supomos que o gradiente de $f$ é Lipschitz. Isto é, existe $L_f \ge 0$ tal que
$$
    \| \nabla f( \vect x ) - \nabla f( \vect y ) \|_2 \le L_f\| \vect x - \vect y \|_2
$$
e que o operador proximal de $\lambda g$, conforme definido abaixo, pode ser calculado facilmente
$$
    \prox_{\lambda g}( \vect x ) := \argmin\limits_{\vect y \in \mathbb R^n} \quad \frac12 \| \vect x - \vect y \|_2^2 + \lambda g( \vect y ).
$$

<h3>1.1. Quadrados Mínimos Regularizado</h3>

Consideremos um problema de quadrados mínimos regularizado da forma
$$
    \min_{\vect x \in \mathbb R^n} \quad \frac12 \| A\vect x - \vect b \|_2^2 + \gamma\| H\vect x \|_1,
$$
com $\gamma \ge 0$ e $H\in\mathbb R^{n\times n}$ ortonormal.

Neste caso, temos
$$
    f( \vect x ) = \frac12 \| A\vect x - \vect b \|_2^2,
$$
$$
    \nabla f( \vect x ) = A^T( A\vect x - \vect b )
$$
e
$$
    \nabla^2 f( \vect x ) = A^TA.
$$

Como a hessiana $\nabla^2 f( \vect x )$ é semi-definida positiva, $f$ é convexa. Além disso, observe que
\begin{align*}
    \| \nabla f( \vect x ) - \nabla f( \vect y ) \|_2 &{}= \| A^TA( \vect x - \vect y ) \|_2\\
        &{}\le \|A^TA\|_2 \| \vect x - \vect y \|_2.
\end{align*}

O proximal de $\lambda g( \vect x ) = \lambda\gamma\| H\vect x \|_1$ é dado por
$$
    \prox_{\lambda g}( \vect x ) = H^T\mathcal S_{\lambda\gamma}( H \vect x ),
$$
onde
$$
    \mathcal S_\alpha( \vect y ) = \begin{cases}
        y_i + \alpha & \text{se } y_i \le -\alpha\\
        0 & \text{se } y_i \in ( -\alpha, \alpha )\\
        y_i - \alpha & \text{se } y_i \ge \alpha
    \end{cases}
$$
é o operador de _soft thresholding_.

<h2>2. Um Algoritmo Básico</h2>

Métodos de gradiente proximal utilizam apenas
- o operador proximal de $\lambda g$

- informações do gradiente de $f$

- o valor da função objetivo $f + g$.

Muitas vezes, para problemas com um grande número $n$ de variáveis, estas informações são tudo que podemos calcular em um tempo razoável.

Por exemplo, para este tipo de problema, o algoritmo de _proximal forward-backward splitting_ (com passo fixo)
$$
    \vect x^{( k + 1 )} = \prox_{\lambda g}\bigl( \vect x^{ ( k ) } - \lambda \nabla f( \vect x^{ ( k ) } ) \bigr)
$$
é convergente se $\lambda > 0$ é suficientemente pequeno (mais precisamente, para $\lambda < 2 / L_f$).

<h3>2.1. Quadrados Mínimos Regularizado</h3>

Para o problema de quadrados mínimos regularizado
$$
    \min_{\vect x \in \mathbb R^n} \quad \frac12 \| A\vect x - \vect b \|_2^2 + \gamma \| H\vect x \|_1.
$$

O método
$$
    \vect x^{( k + 1 )} = \prox_{\lambda g}\bigl( \vect x^{ ( k ) } - \lambda \nabla f( \vect x^{ ( k ) } ) \bigr)
$$
assume, portanto, a forma
$$
    \vect x^{( k + 1 )} = H^T\mathcal S_{\lambda\gamma}\Bigl( H\bigl( \vect x^{ ( k ) } - \lambda A^T( A\vect x^{ ( k ) } - \vect b ) \bigr) \Bigr).
$$

Este algoritmo tem a seguinte estrutura:
- Caminhe na direção de máxima descida

- Calcule a transformação esparsificante

- Faça o _soft-thresholding_

- Calcule a transformação inversa

<h3>2.2. Aplicação</h3>

Neste exemplo,  os dados $\vect b$ serão os pixels de uma versão borrada e ruidosa de uma imagem desconhecida e desejamos obter a imagem original a partir da sua versão borrada.

O operador $A$ é tal que se $\vect x$ é a imagem original, então $A\vect x$ é a versão borrada. Esta operação $\vect x \mapsto A\vect x$ é obtida através da _convolução_ da imagem com uma imagem constante, chamada de _kernel_ da convolução.

A adjunta $A^T$ da convolução é a _correlação_. Ambas operações estão implementadas no módulo `scipy.signal` através das funções `convolve` e `correlate`.

Principiamos por nossa imagem inicial, que é a foto de uma astronauta. Neste exemplo, usaremos uma versão monocromática da imagem original colorida. Esta imagem está disponível no módulo `skimage`:

```python
import pywt

wavs = pywt.wavelist()
wav = wavs[ 0 ]
wav = 'db5'

def H( x ):
    return pywt.coeffs_to_array( pywt.wavedec2( x, wav, mode = 'periodization' ) )

def HT( c ):
    return pywt.waverec2( pywt.array_to_coeffs( *c, output_format = 'wavedec2' ), wav, mode = 'periodization' )
```

Testa  $\vect y^T H \vect x = \vect x^T H^T\vect y$ e se $H^TH = I$:

```python
def prox( x, alpha ):

    Hx = H( x )
    return HT( ( pywt.threshold( Hx[ 0 ], alpha ), Hx[ 1 ] ) )
```

```python
def ista( x, grad, prox, step, gamma, niter ):

    iters = [ x ]

    for _ in range( niter ):

        x = prox( x - step * grad( x ), step * gamma )
        iters.append( x )

    return iters
```

## 3. Um Algoritmo Acelerado

O algoritmo FISTA tem a seguinte forma
$$
\begin{align*}
    \vect y^{( k )} &{}= \prox_{\lambda g}\bigl( \vect x^{( k )} - \lambda \nabla f( \vect x^{( k )} ) \bigr)\\
    t_{k + 1} &{}= \frac{1 + \sqrt{1 + 4t_k^2}}2\\
    \vect x^{( k + 1 )} &{}= \vect y^{( k )} + \frac{t_k - 1}{t_{k + 1}}( \vect y^{( k )} - \vect y^{( k - 1 )} ).
\end{align*}
$$

```python
def fista( x, grad, prox, step, gamma, niter ):

    iters = [ x ]

    y_prev = x
    t = 1

    for _ in range( niter ):

        y = prox( x - step * grad( x ), step * gamma )

        t_prev = t
        t = 0.5 * ( 1 + ( 1 + 4 * t ** 2 ) ** 0.5 )

        x = y + ( t_prev - 1 ) / t * ( y - y_prev )

        y_prev = y

        iters.append( x )

    return iters
```

```python
iters_f = fista( b, grad, prox, 1.0, gamma, 10 * niter )
```

```python
f_vals_f = [ f( x ) + gamma * np.abs( H( x )[ 0 ] ).sum() for x in iters_f[ : niter + 1 ] ]

f_star = f( iters_f[ -1 ] ) + gamma * np.abs( H( iters_f[ -1 ] )[ 0 ] ).sum()
d_vals_f = [ f - f_star for f in f_vals_f ]
d_vals_i = [ f - f_star for f in f_vals_i ]

err_f = [ np.abs( img - x ).sum() for x in iters_f[ : niter + 1 ]  ]
```

```python
pp.figure( figsize = ( 16, 32 ) )

pp.subplot( 4, 2, 1 )
pp.imshow( img, cmap = pp.cm.gray )
pp.title( 'Original' )
pp.axis( 'off' )

pp.subplot( 4, 2, 2 )
pp.imshow( iters_f[ -1 ], cmap = pp.cm.gray )
pp.title( 'Reconstrução' )
pp.axis( 'off' )

pp.subplot( 4, 1, 2 )
pp.semilogy( d_vals_i )
pp.semilogy( d_vals_f )
pp.xlabel( 'Iteração' )
pp.ylabel( '$F( \\boldsymbol x_k ) - F^*$' )
pp.legend( [ 'ISTA', 'FISTA' ] )
pp.axis( 'tight' )

pp.subplot( 4, 2, 5 )
pp.imshow( iters_i[ -1 ], cmap = pp.cm.gray )
pp.title( 'ISTA' )
pp.axis( 'off' )

pp.subplot( 4, 2, 6 )
pp.imshow( iters_f[ -1 ], cmap = pp.cm.gray )
pp.title( 'FISTA' )
pp.axis( 'off' )

pp.subplot( 4, 1, 4 )
pp.semilogy( err_i )
pp.semilogy( err_f )
pp.xlabel( 'Iteração' )
pp.ylabel( 'Erro' )
pp.legend( [ 'ISTA', 'FISTA' ] )
pp.axis( 'tight' )

pp.show()
```
